En Exam_Score se encuentra un valor atipico en el valor maximo ya que esta marcado como 101 y el maximo deberia ser 100, asi que puede representar un valor atipico o un error de captura al trabajar el dataset.

```python
issues = []

if df['Exam_Score'].max() > 100:
    issues.append({
        "Problema detectado": "Valor fuera de rango / Extremo",
        "Variable afectada": "Exam_Score",
        "Evidencia": f"El valor máximo de Exam_Score es {df['Exam_Score'].max()}, cuando el máximo esperado es 100.",
        "Tratamiento propuesto": "Corregir o imputar el valor. Si es un error de entrada, se podría reemplazar por 100 o considerar como un valor atípico a tratar."
    })
```

```python
missing_values_cols = df.isnull().sum()
missing_values_cols = missing_values_cols[missing_values_cols > 0]
for col, count in missing_values_cols.items():
    issues.append({
        "Problema detectado": "Valores faltantes",
        "Variable afectada": col,
        "Evidencia": f"{count} valores nulos ({count/len(df)*100:.2f}%)",
        "Tratamiento propuesto": "Imputación (media, mediana, moda) o eliminación de filas/columnas, dependiendo del contexto y la proporción de datos faltantes."
    })
```

```python
if issues:
    display(pd.DataFrame(issues))
else:
    print("No se detectaron problemas de calidad de datos significativos más allá de los ya mencionados (0 duplicados).")
```

```python
df.describe(include='object').head()
```

```python
categorical_cols_with_na = ['Teacher_Quality', 'Parental_Education_Level', 'Distance_from_Home']

for col in categorical_cols_with_na:
    # Se aplica .str.strip() para eliminar espacios en blanco al inicio/final
    # y .str.title() o .str.lower() para estandarizar la capitalización si es necesario.
    # Para este dataset, una simple limpieza de espacios y unificación de caso es suficiente.
    if df[col].dtype == 'object':
        df[col] = df[col].str.strip().replace('', np.nan) # Reemplazar cadenas vacías con NaN
        # Opcional: Estandarizar capitalización. Decido no hacerlo para no alterar las etiquetas originales si no es estrictamente necesario.
        # df[col] = df[col].str.title() # o .str.lower()
        print(f'Valores únicos estandarizados para {col}: {df[col].dropna().unique()}')
```

```python
if df.isnull().sum().sum() == 0:
    print('Todos los valores faltantes han sido tratados.')
else:
    print('Advertencia: Todavía existen valores faltantes en el dataset.')
```

```python
print("--- Medidas Descriptivas de Exam_Score ---")
display(df['Exam_Score'].describe())

# Configuración de los gráficos
plt.figure(figsize=(15, 6))

# Histograma de Exam_Score
plt.subplot(1, 2, 1) # 1 fila, 2 columnas, primer gráfico
sns.histplot(df['Exam_Score'], kde=True, bins=20, color='skyblue')
plt.title('Distribución de Exam_Score')
plt.xlabel('Puntuación del Examen')
plt.ylabel('Frecuencia')
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Diagrama de Caja de Exam_Score
plt.subplot(1, 2, 2) # 1 fila, 2 columnas, segundo gráfico
sns.boxplot(y=df['Exam_Score'], color='lightcoral')
plt.title('Diagrama de Caja de Exam_Score')
plt.ylabel('Puntuación del Examen')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

### Interpretación de la Distribución de `Exam_Score`

**Observaciones:**

*   **Estadísticas Descriptivas:**
    *   La puntuación media es aproximadamente `75.07`, lo que sugiere un rendimiento académico por encima del promedio si la escala es de 0 a 100.
    *   La desviación estándar es de aproximadamente `14.39`, indicando una variabilidad moderada en las puntuaciones. No todos los estudiantes tienen un rendimiento similar.
    *   El rango de puntuaciones va desde `50` hasta `100`, con cuartiles que muestran la concentración de datos.

*   **Histograma:**
    *   La distribución del `Exam_Score` parece ser aproximadamente simétrica y unimodal, con la mayoría de los estudiantes obteniendo puntuaciones en el rango medio.
    *   La curva de densidad (KDE) sugiere una distribución normal o cercana a la normal, centrada alrededor de la media.

*   **Diagrama de Caja:**
    *   La caja central del diagrama abarca el 50% de los datos (entre el primer y tercer cuartil), mostrando la mayor concentración de puntuaciones.
    *   Los bigotes del diagrama se extienden sin mostrar puntos individuales más allá de ellos. Esto indica que, después de la limpieza de datos donde se truncó el valor máximo a 100, no hay valores atípicos evidentes que estén significativamente alejados del resto de los datos en `Exam_Score`.

### Impacto en el Modelado Posterior

**Conclusiones para el Modelado:**

*   **Distribución Apropiada:** La distribución de `Exam_Score` parece ser adecuada para un modelo de regresión. No se observan sesgos extremos ni múltiples modas que pudieran complicar la modelización.
*   **Ausencia de Valores Atípicos Extremos:** Gracias a la limpieza previa, no hay valores atípicos que requieran un tratamiento especial adicional. Esto asegura que el modelo no se vea indebidamente influenciado por datos anómalos.
*   **Escala de la Variable:** La variable ya está en una escala numérica que es directamente interpretable (0-100), lo que facilita su uso como variable objetivo sin necesidad de transformaciones complejas en una primera fase.
